# TruthfulQA

```python
import pandas as pd
import json
```

```python
def load_csv(file_path, is_gzip=False):
    # input file is in csv format, can be loaded by pandas
    # required columns: [Question] only
    list_data = []
    with open(file_path, "r") as f:
        df = pd.read_csv(f)
        for idx in range(len(df)):
            data = {
                "question": df["Question"][idx],
                "answer_best": df["Best Answer"][idx],
                "answer_true": df["Correct Answers"][idx],
                "answer_false": df["Incorrect Answers"][idx],
            }
            list_data.append(data)

    return list_data
```

```python
list_data_dict = load_csv(file_path="data/tfqa/TruthfulQA.csv")
```

```python
import random

# 1. Setup
all_indices = list(range(len(list_data_dict)))
random.seed(42)
random.shuffle(all_indices)

k = 1              # Number of sets
samples_per_set = 300 

# 2. Extract non-overlapping sets
sets = []
for i in range(k):
    start = i * samples_per_set
    end = start + samples_per_set
    
    # Slice the shuffled list
    subset = all_indices[start:end]
    sets.append(subset)

# Verification
fold_0 = sets[0]

print(f"Fold 0 size: {len(fold_0)}")
```

```python
import random

# 1. Setup
all_indices = list(range(len(list_data_dict)))
random.seed(42)
random.shuffle(all_indices)

k = 2              # Number of sets
samples_per_set = 150

# 2. Extract non-overlapping sets
sets = []
for i in range(k):
    start = i * samples_per_set
    end = start + samples_per_set
    
    # Slice the shuffled list
    subset = all_indices[start:end]
    sets.append(subset)

# Verification
fold_0 = sets[0]
fold_1 = sets[1]
overlap = set(fold_0).intersection(set(fold_1))

print(f"Fold 0 size: {len(fold_0)}")
print(f"Fold 1 size: {len(fold_1)}")
print(f"Overlap: {overlap}") # Should be set()
```

```python
data_fold_0 = []
for idx, item in enumerate(list_data_dict):
    if idx in fold_0:
        data_fold_0.append(item)
print("len", len(data_fold_0))
with open("data/tfqa/tfqa_val_fold_0_300.json", "w") as f:
    json.dump(data_fold_0, f, indent=4)
```

# TriviaQA

```python
from datasets import load_dataset, Dataset
```

```python
dataset = load_dataset("trivia_qa", "unfiltered.nocontext")["train"]
```

```python
list_data = list(dataset["question"])
labels = list(dataset["answer"])
```

```python
import random

# 1. Setup
all_indices = list(range(len(list_data)))
random.seed(62)
random.shuffle(all_indices)

k = 2              # Number of sets
samples_per_set = 2000 

# 2. Extract non-overlapping sets
sets = []
for i in range(k):
    start = i * samples_per_set
    end = start + samples_per_set
    
    # Slice the shuffled list
    subset = all_indices[start:end]
    sets.append(subset)

# Verification
fold_0 = sets[0]
fold_1 = sets[1]
overlap = set(fold_0).intersection(set(fold_1))

print(f"Fold 0 size: {len(fold_0)}")
print(f"Fold 1 size: {len(fold_1)}")
print(f"Overlap: {overlap}") # Should be set()
```

```python
list_data_fold_0 = []
labels_fold_0 = []
for idx in fold_0:
    list_data_fold_0.append(list_data[idx])
    labels_fold_0.append(labels[idx])

print("len", len(list_data_fold_0))


dataset_fold_0 = Dataset.from_list(
    [
        {"question": question, "answer": answer}
        for question, answer in zip(list_data_fold_0, labels_fold_0)
    ]
)


dataset_fold_0.save_to_disk("data/triviaqa/triviaqa_val_fold_0")
```

```python
list_data_fold_1 = []
labels_fold_1 = []
for idx in fold_1:
    list_data_fold_1.append(list_data[idx])
    labels_fold_1.append(labels[idx])

print("len", len(list_data_fold_1))


dataset_fold_1 = Dataset.from_list(
    [
        {"question": question, "answer": answer}
        for question, answer in zip(list_data_fold_1, labels_fold_1)
    ]
)


dataset_fold_1.save_to_disk("data/triviaqa/triviaqa_val_fold_1")
```
